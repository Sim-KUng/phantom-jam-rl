```python
import gymnasium as gym
from gymnasium.wrappers import RecordVideo
import highway_env

print("🎥 비디오 녹화 테스트를 시작합니다...")

# 1. 환경 생성 시 render_mode를 'rgb_array'로 설정 (중요!)
env = gym.make("highway-v0", render_mode="rgb_array")

# 2. 영상을 구글 드라이브의 특정 폴더에 저장하도록 세팅
# (구글 드라이브 최상단에 phantom-jam-videos 폴더가 자동 생성됩니다)
video_path = '/content/drive/MyDrive/colab/Learning_Sharing/phantom-jam-videos'
env = RecordVideo(env, video_folder=video_path, episode_trigger=lambda e: True)

# 3. 환경 가볍게 세팅
env.unwrapped.configure({
    "lanes_count": 2,
    "vehicles_count": 15,
    "duration": 20 # 짧게 20프레임만 녹화
})

env.reset()

# 4. 시뮬레이션 루프
done = truncated = False
while not (done or truncated):
    obs, reward, done, truncated, info = env.step(1) # IDLE 액션

env.close()

print(f"✅ 테스트 완료! 구글 드라이브의 '{video_path}' 폴더를 확인해 보세요.")
```
